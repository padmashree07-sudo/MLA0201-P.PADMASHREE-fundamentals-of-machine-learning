```python
"""Experiment 1: FIND-S algorithm (most specific hypothesis) with matplotlib."""
import csv
import matplotlib.pyplot as plt

# ---- Load data ----------------------------------------------------------
# Option A: read from a CSV file (last column = target "Yes"/"No")
def load_csv(path):
    with open(path) as f:
        rows = list(csv.reader(f))
    return rows[0], rows[1:]

# Option B: built-in EnjoySport dataset (used if no CSV is available)
header = ["Sky", "AirTemp", "Humidity", "Wind", "Water", "Forecast", "EnjoySport"]
data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same",   "Yes"],
    ["Sunny", "Warm", "High",   "Strong", "Warm", "Same",   "Yes"],
    ["Rainy", "Cold", "High",   "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High",   "Strong", "Cool", "Change", "Yes"],
]
# header, data = load_csv("enjoysport.csv")   # <- uncomment to use your CSV

# ---- FIND-S -------------------------------------------------------------
n_attr = len(header) - 1
hypothesis = None
history = []          # hypothesis after each training example

for i, row in enumerate(data, 1):
    x, label = row[:-1], row[-1]
    if label.lower() == "yes":
        if hypothesis is None:
            hypothesis = x.copy()          # first positive example
        else:
            for j in range(n_attr):
                if hypothesis[j] != x[j]:
                    hypothesis[j] = "?"    # generalise
    # negative examples are ignored by FIND-S
    snapshot = hypothesis.copy() if hypothesis else ["0"] * n_attr
    history.append(snapshot)
    print(f"Example {i} ({label}): h = {snapshot}")

print("\nMost specific hypothesis:", history[-1])

# ---- Plots --------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))

# (a) hypothesis table after each example
ax = axes[0]
ax.axis("off")
ax.set_title("Hypothesis after each training example")
tbl = ax.table(cellText=[[f"{i+1}"] + h for i, h in enumerate(history)],
               colLabels=["Ex."] + header[:-1], loc="center")
tbl.auto_set_font_size(False)
tbl.set_fontsize(9)
tbl.scale(1, 1.8)

# (b) how many attributes are still specific
ax = axes[1]
specific = [sum(1 for v in h if v not in ("?", "0")) for h in history]
ax.plot(range(1, len(history) + 1), specific, marker="o", color="tab:blue")
ax.set_xticks(range(1, len(history) + 1))
ax.set_xlabel("Training example #")
ax.set_ylabel("# attributes still specific")
ax.set_title("FIND-S: generalisation of the hypothesis")
ax.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
"""Experiment 2: Candidate-Elimination algorithm with matplotlib."""
import csv
import matplotlib.pyplot as plt

def load_csv(path):
    with open(path) as f:
        rows = list(csv.reader(f))
    return rows[0], rows[1:]

header = ["Sky", "AirTemp", "Humidity", "Wind", "Water", "Forecast", "EnjoySport"]
data = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same",   "Yes"],
    ["Sunny", "Warm", "High",   "Strong", "Warm", "Same",   "Yes"],
    ["Rainy", "Cold", "High",   "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High",   "Strong", "Cool", "Change", "Yes"],
]
# header, data = load_csv("enjoysport.csv")   # <- uncomment to use your CSV

n = len(header) - 1
domains = [sorted({row[i] for row in data}) for i in range(n)]

def covers(h, x):
    return all(a == "?" or a == b for a, b in zip(h, x))

def more_general(h1, h2):          # is h1 >= h2 ?
    return all(a == "?" or a == b or b == "0" for a, b in zip(h1, h2))

S = ["0"] * n
G = [["?"] * n]
size_S, size_G = [], []

for t, row in enumerate(data, 1):
    x, label = row[:-1], row[-1].lower()
    if label == "yes":
        G = [g for g in G if covers(g, x)]
        for i in range(n):
            if S[i] == "0":
                S[i] = x[i]
            elif S[i] != x[i]:
                S[i] = "?"
    else:
        new_G = []
        for g in G:
            if not covers(g, x):
                new_G.append(g)
            else:
                for i in range(n):
                    if g[i] == "?":
                        for v in domains[i]:
                            if v != x[i]:
                                h = g.copy()
                                h[i] = v
                                if more_general(h, S):
                                    new_G.append(h)
        # keep only maximally general, unique hypotheses
        unique = []
        for h in new_G:
            if h not in unique:
                unique.append(h)
        G = [h for h in unique
             if not any(h != o and more_general(o, h) for o in unique)]

    size_S.append(sum(1 for v in S if v == "?"))   # '?' count in S
    size_G.append(len(G))
    print(f"\nAfter example {t} ({label}):")
    print("  S =", S)
    print("  G =", G)

print("\nFinal specific boundary S:", S)
print("Final general boundary  G:", G)

# ---- Plots --------------------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(14, 4.5))
steps = range(1, len(data) + 1)

ax = axes[0]
ax.plot(steps, size_S, marker="o", label="# '?' in S (S generalises)")
ax.plot(steps, size_G, marker="s", label="|G| (number of general hypotheses)")
ax.set_xticks(list(steps))
ax.set_xlabel("Training example #")
ax.set_title("Version space boundaries")
ax.legend()
ax.grid(alpha=0.3)

ax = axes[1]
ax.axis("off")
ax.set_title("Final boundaries")
cells = [["S"] + S] + [[f"G{i+1}"] + g for i, g in enumerate(G)]
tbl = ax.table(cellText=cells, colLabels=[""] + header[:-1], loc="center")
tbl.auto_set_font_size(False)
tbl.set_fontsize(9)
tbl.scale(1, 2)

plt.tight_layout()
plt.show()
```

```python
"""Experiment 5: K-Nearest Neighbours from scratch (Iris) with matplotlib."""
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter
from matplotlib.colors import ListedColormap
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

iris = load_iris()
X, y = iris.data, iris.target
X_tr, X_te, y_tr, y_te = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

def knn_predict_one(x, X_train, y_train, k):
    dist = np.sqrt(((X_train - x) ** 2).sum(axis=1))      # Euclidean distance
    nearest = np.argsort(dist)[:k]
    return Counter(y_train[nearest]).most_common(1)[0][0]  # majority vote

def knn_predict(X_test, X_train, y_train, k):
    return np.array([knn_predict_one(x, X_train, y_train, k) for x in X_test])

# ---- Accuracy for k = 5 -------------------------------------------------
k = 5
pred = knn_predict(X_te, X_tr, y_tr, k)
print(f"K = {k}  ->  accuracy = {np.mean(pred == y_te) * 100:.2f}%")

# ---- Accuracy for different k -------------------------------------------
ks = range(1, 26)
accs = [np.mean(knn_predict(X_te, X_tr, y_tr, kk) == y_te) for kk in ks]
best_k = ks[int(np.argmax(accs))]
print("Best k:", best_k, "with accuracy", round(max(accs), 4))

# ---- Plots --------------------------------------------------------------
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# (a) accuracy vs k
axes[0].plot(list(ks), accs, marker="o")
axes[0].axvline(best_k, color="red", ls="--", label=f"best k = {best_k}")
axes[0].set_xlabel("k"); axes[0].set_ylabel("Test accuracy")
axes[0].set_title("Accuracy vs k"); axes[0].legend(); axes[0].grid(alpha=0.3)

# (b) decision boundary using petal length & petal width
f1, f2 = 2, 3
Xp_tr, Xp_te = X_tr[:, [f1, f2]], X_te[:, [f1, f2]]
xx, yy = np.meshgrid(np.arange(0.5, 7.5, 0.05), np.arange(0, 3, 0.03))
grid = np.c_[xx.ravel(), yy.ravel()]
zz = knn_predict(grid, Xp_tr, y_tr, k).reshape(xx.shape)

axes[1].contourf(xx, yy, zz, alpha=0.3, cmap=ListedColormap(["#FFAAAA", "#AAFFAA", "#AAAAFF"]))
for c, name in enumerate(iris.target_names):
    axes[1].scatter(Xp_tr[y_tr == c, 0], Xp_tr[y_tr == c, 1], label=name, edgecolor="k", s=30)
axes[1].set_xlabel(iris.feature_names[f1]); axes[1].set_ylabel(iris.feature_names[f2])
axes[1].set_title(f"KNN decision boundary (k={k})"); axes[1].legend()

# (c) confusion matrix
ConfusionMatrixDisplay(confusion_matrix(y_te, pred),
                       display_labels=iris.target_names).plot(ax=axes[2], colorbar=False)
axes[2].set_title(f"Confusion matrix (k={k})")

plt.tight_layout()
plt.show()
```

```python
# Experiment 10: Expectation-Maximization (EM) algorithm
# GaussianMixture uses the EM algorithm to find the hidden groups in the data
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# 1. Make data from two groups
np.random.seed(0)
group1 = np.random.normal(-2, 1, 200)
group2 = np.random.normal(4, 1.5, 300)
X = np.concatenate([group1, group2]).reshape(-1, 1)

# 2. Run EM (2 components)
model = GaussianMixture(n_components=2, random_state=0)
model.fit(X)

# 3. Print results
print("Means:", model.means_.ravel())
print("Std devs:", np.sqrt(model.covariances_.ravel()))
print("Weights:", model.weights_)

# 4. Plot
labels = model.predict(X)
plt.scatter(X[:, 0], np.zeros(len(X)), c=labels, cmap="coolwarm", alpha=0.5)
plt.hist(X[:, 0], bins=30, density=True, alpha=0.3)
plt.title("EM Algorithm - Two Clusters Found")
plt.show()
```

```python
# Experiment 11: Credit Score Classification (Poor / Standard / Good)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

# 1. Make data (to use your own file: df = pd.read_csv("credit.csv"))
np.random.seed(0)
n = 600
df = pd.DataFrame({
    "Age": np.random.randint(18, 65, n),
    "Income": np.random.randint(20000, 120000, n),
    "Debt": np.random.randint(0, 50000, n),
    "LatePayments": np.random.randint(0, 15, n),
})
score = df["Income"] / 1000 - df["Debt"] / 1000 - df["LatePayments"] * 5
df["CreditScore"] = pd.cut(score, bins=[-100, 20, 60, 200], labels=["Poor", "Standard", "Good"])

# 2. Split data
X = df.drop("CreditScore", axis=1)
y = df["CreditScore"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# 3. Train model
model = RandomForestClassifier(random_state=1)
model.fit(X_train, y_train)

# 4. Test model
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))

# 5. Plots
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ConfusionMatrixDisplay.from_predictions(y_test, pred, ax=ax[0])
ax[0].set_title("Confusion Matrix")
ax[1].barh(X.columns, model.feature_importances_)
ax[1].set_title("Feature Importance")
plt.tight_layout()
plt.show()
```

```python
# Experiment 12: Iris Flower Classification using KNN
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

iris = load_iris()
X, y = iris.data, iris.target
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = KNeighborsClassifier(n_neighbors=5)
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))

# Try different k values
k_values = range(1, 21)
accuracies = []
for k in k_values:
    m = KNeighborsClassifier(n_neighbors=k).fit(X_train, y_train)
    accuracies.append(m.score(X_test, y_test))

fig, ax = plt.subplots(1, 3, figsize=(16, 4))
ax[0].scatter(X[:, 2], X[:, 3], c=y)
ax[0].set_xlabel("Petal length"); ax[0].set_ylabel("Petal width")
ax[0].set_title("Iris data")
ax[1].plot(k_values, accuracies, marker="o")
ax[1].set_xlabel("k"); ax[1].set_ylabel("Accuracy")
ax[1].set_title("Accuracy vs k")
ConfusionMatrixDisplay.from_predictions(y_test, pred, display_labels=iris.target_names, ax=ax[2])
ax[2].set_title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# Experiment 13: Car Price Prediction
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

# 1. Make data (to use your own file: df = pd.read_csv("car.csv"))
np.random.seed(0)
n = 200
df = pd.DataFrame({
    "Age": np.random.randint(1, 15, n),
    "KmDriven": np.random.randint(5000, 150000, n),
    "EngineCC": np.random.choice([800, 1000, 1200, 1500, 2000], n),
})
df["Price"] = 800000 - df["Age"] * 30000 - df["KmDriven"] * 2 + df["EngineCC"] * 150 + np.random.randn(n) * 20000

# 2. Split data
X = df[["Age", "KmDriven", "EngineCC"]]
y = df["Price"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# 3. Train model
model = LinearRegression()
model.fit(X_train, y_train)

# 4. Test model
pred = model.predict(X_test)
print("R2 score:", r2_score(y_test, pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred)))

# 5. Plots
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].scatter(y_test, pred)
ax[0].plot([y.min(), y.max()], [y.min(), y.max()], "r--")
ax[0].set_xlabel("Actual Price"); ax[0].set_ylabel("Predicted Price")
ax[0].set_title("Actual vs Predicted")
ax[1].scatter(df["Age"], df["Price"], alpha=0.5)
ax[1].set_xlabel("Car Age (years)"); ax[1].set_ylabel("Price")
ax[1].set_title("Price vs Age")
plt.tight_layout()
plt.show()
```

```python
# Experiment 14: House Price Prediction
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_squared_error, r2_score

# 1. Make data (to use your own file: df = pd.read_csv("house.csv"))
np.random.seed(1)
n = 300
df = pd.DataFrame({
    "Area": np.random.randint(500, 3500, n),
    "Bedrooms": np.random.randint(1, 6, n),
    "Age": np.random.randint(0, 40, n),
})
df["Price"] = df["Area"] * 3000 + df["Bedrooms"] * 200000 - df["Age"] * 15000 + np.random.randn(n) * 100000

# 2. Split data
X = df[["Area", "Bedrooms", "Age"]]
y = df["Price"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# 3. Train model
model = RandomForestRegressor(random_state=1)
model.fit(X_train, y_train)

# 4. Test model
pred = model.predict(X_test)
print("R2 score:", r2_score(y_test, pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, pred)))

# 5. Plots
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].scatter(y_test, pred)
ax[0].plot([y.min(), y.max()], [y.min(), y.max()], "r--")
ax[0].set_xlabel("Actual Price"); ax[0].set_ylabel("Predicted Price")
ax[0].set_title("Actual vs Predicted")
ax[1].barh(X.columns, model.feature_importances_)
ax[1].set_title("Feature Importance")
plt.tight_layout()
plt.show()
```

```python
# Experiment 16: Compare different Classification Algorithms
import matplotlib.pyplot as plt
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

data = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.3, random_state=42)

# Scale the data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "KNN": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(random_state=1),
    "SVM": SVC(),
    "Random Forest": RandomForestClassifier(random_state=1),
}

names, accs = [], []
print(f"{'Model':22s}{'Acc':>8s}{'Prec':>8s}{'Recall':>8s}{'F1':>8s}")
for name, model in models.items():
    model.fit(X_train, y_train)
    pred = model.predict(X_test)
    acc = accuracy_score(y_test, pred)
    print(f"{name:22s}{acc:8.3f}{precision_score(y_test, pred):8.3f}"
          f"{recall_score(y_test, pred):8.3f}{f1_score(y_test, pred):8.3f}")
    names.append(name)
    accs.append(acc)

plt.figure(figsize=(9, 4))
bars = plt.bar(names, accs, color="skyblue")
plt.bar_label(bars, fmt="%.3f")
plt.ylim(0.85, 1.0)
plt.ylabel("Accuracy")
plt.title("Comparison of Classification Algorithms")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

```python
# Experiment 17: Mobile Price Range Prediction (0=low, 1=medium, 2=high, 3=very high)
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

# 1. Make data (to use your own file: df = pd.read_csv("mobile.csv"))
np.random.seed(2)
n = 600
df = pd.DataFrame({
    "RAM_GB": np.random.choice([2, 4, 6, 8, 12], n),
    "Storage_GB": np.random.choice([32, 64, 128, 256], n),
    "Battery_mAh": np.random.randint(2500, 6000, n),
    "Camera_MP": np.random.randint(8, 108, n),
})
score = df["RAM_GB"] * 3 + df["Storage_GB"] / 16 + df["Battery_mAh"] / 500 + df["Camera_MP"] / 10
df["PriceRange"] = pd.qcut(score, 4, labels=[0, 1, 2, 3]).astype(int)

# 2. Split data
X = df.drop("PriceRange", axis=1)
y = df["PriceRange"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# 3. Train model
model = RandomForestClassifier(random_state=1)
model.fit(X_train, y_train)

# 4. Test model
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))

# 5. Plots
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ConfusionMatrixDisplay.from_predictions(y_test, pred, ax=ax[0])
ax[0].set_title("Confusion Matrix")
ax[1].barh(X.columns, model.feature_importances_)
ax[1].set_title("Feature Importance")
plt.tight_layout()
plt.show()
```

```python
# Experiment 18: Perceptron based IRIS classification
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score

iris = load_iris()
X = iris.data[:, [2, 3]]          # petal length, petal width
y = iris.target

# Scaling helps the Perceptron learn faster
X = StandardScaler().fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

model = Perceptron(max_iter=1000, random_state=1)
model.fit(X_train, y_train)
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))

# Draw decision regions
xx, yy = np.meshgrid(np.linspace(-2, 2.5, 300), np.linspace(-2, 2.5, 300))
zz = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)

plt.contourf(xx, yy, zz, alpha=0.3)
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolor="k")
plt.xlabel("Petal length (scaled)"); plt.ylabel("Petal width (scaled)")
plt.title("Perceptron Decision Regions (Iris)")
plt.show()
```

```python
# Experiment 19: Naive Bayes for Bank Loan Prediction
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, ConfusionMatrixDisplay

# 1. Make data (to use your own file: df = pd.read_csv("loan.csv"))
np.random.seed(3)
n = 500
df = pd.DataFrame({
    "Age": np.random.randint(21, 65, n),
    "Income": np.random.randint(20000, 120000, n),
    "CreditScore": np.random.randint(300, 850, n),
    "LoanAmount": np.random.randint(5000, 80000, n),
})
chance = (df["CreditScore"] - 300) / 550 + df["Income"] / 120000 - df["LoanAmount"] / 80000
df["Approved"] = (chance > 0.7).astype(int)     # 1 = approved, 0 = rejected

# 2. Split data
X = df.drop("Approved", axis=1)
y = df["Approved"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)

# 3. Train model
model = GaussianNB()
model.fit(X_train, y_train)

# 4. Test model
pred = model.predict(X_test)
print("Accuracy:", accuracy_score(y_test, pred))

# 5. Plots
fig, ax = plt.subplots(1, 2, figsize=(11, 4))
ax[0].scatter(df["CreditScore"], df["Income"], c=df["Approved"], cmap="coolwarm", alpha=0.6)
ax[0].set_xlabel("Credit Score"); ax[0].set_ylabel("Income")
ax[0].set_title("Loan Approved (red) vs Rejected (blue)")
ConfusionMatrixDisplay.from_predictions(y_test, pred, display_labels=["Rejected", "Approved"], ax=ax[1])
ax[1].set_title("Confusion Matrix")
plt.tight_layout()
plt.show()
```

```python
# Experiment 20: Future Sales Prediction
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# 1. Make data: 36 months of sales that slowly go up
np.random.seed(4)
months = np.arange(1, 37).reshape(-1, 1)
sales = 1000 + months[:, 0] * 40 + np.random.randn(36) * 80

# 2. Train model
model = LinearRegression()
model.fit(months, sales)

# 3. Predict next 12 months
future = np.arange(37, 49).reshape(-1, 1)
future_sales = model.predict(future)
print("Predicted sales for next 12 months:")
for m, s in zip(future[:, 0], future_sales):
    print(f"Month {m}: {s:.0f}")

# 4. Plot
plt.plot(months, sales, label="Past sales")
plt.plot(months, model.predict(months), "g--", label="Trend line")
plt.plot(future, future_sales, "r-o", label="Predicted sales")
plt.xlabel("Month"); plt.ylabel("Sales")
plt.title("Future Sales Prediction")
plt.legend()
plt.show()
```
