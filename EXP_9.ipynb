```python
# Experiment 9: Compare Linear and Polynomial Regression
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

# 1. Make curved data: y = x^2 + noise
np.random.seed(0)
X = np.sort(np.random.rand(80, 1) * 6 - 3, axis=0)
y = X[:, 0] ** 2 + np.random.randn(80) * 0.8

# 2. Linear regression
linear = LinearRegression()
linear.fit(X, y)
linear_pred = linear.predict(X)

# 3. Polynomial regression (degree 2)
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)
poly_model = LinearRegression()
poly_model.fit(X_poly, y)
poly_pred = poly_model.predict(X_poly)

# 4. Compare
print("Linear R2     :", r2_score(y, linear_pred))
print("Polynomial R2 :", r2_score(y, poly_pred))

# 5. Plot
plt.scatter(X, y, label="Data")
plt.plot(X, linear_pred, "r-", label="Linear")
plt.plot(X, poly_pred, "g-", label="Polynomial (degree 2)")
plt.title("Linear vs Polynomial Regression")
plt.legend()
plt.show()
```
