```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# ---------------------------------------
# 1. Dataset
# XOR problem
# ---------------------------------------

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 1, 1, 0])


# ---------------------------------------
# 2. Create Training and Testing Data
# ---------------------------------------

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.5,
    random_state=42
)


# ---------------------------------------
# 3. Create Artificial Neural Network
# ---------------------------------------

model = MLPClassifier(
    hidden_layer_sizes=(4,),
    activation='relu',
    solver='adam',
    learning_rate_init=0.1,
    max_iter=5000,
    random_state=42
)


# ---------------------------------------
# 4. Train the Neural Network
# ---------------------------------------

model.fit(X_train, y_train)


# ---------------------------------------
# 5. Prediction
# ---------------------------------------

y_pred = model.predict(X_test)

print("Actual Values    :", y_test)
print("Predicted Values :", y_pred)

accuracy = accuracy_score(y_test, y_pred)

print("\nAccuracy:", accuracy * 100, "%")


# ---------------------------------------
# 6. Test XOR Inputs
# ---------------------------------------

predictions = model.predict(X)

print("\nXOR Predictions:")

for inputs, prediction in zip(X, predictions):
    print(inputs, "=>", prediction)


# ---------------------------------------
# 7. Matplotlib Visualization
# ---------------------------------------

plt.figure(figsize=(7, 5))

plt.scatter(
    X[:, 0],
    X[:, 1],
    c=predictions,
    s=200
)

for i, prediction in enumerate(predictions):
    plt.text(
        X[i, 0] + 0.03,
        X[i, 1] + 0.03,
        "Output = " + str(prediction),
        fontsize=10
    )

plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.title("Artificial Neural Network - XOR Classification")

plt.xticks([0, 1])
plt.yticks([0, 1])
plt.grid(True)

plt.show()


# ---------------------------------------
# 8. Loss Curve
# ---------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    model.loss_curve_,
    linewidth=2
)

plt.xlabel("Iterations")
plt.ylabel("Loss")
plt.title("Backpropagation Training Loss")

plt.grid(True)

plt.show()
```
